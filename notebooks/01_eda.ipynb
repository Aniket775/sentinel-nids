# Network Intrusion Detection System (NIDS) — Exploratory Data Analysis (EDA)
## Dataset: CICIDS2017 Benchmark
This notebook demonstrates exploratory data analysis on the Canadian Institute for Cybersecurity (CIC) CICIDS2017 flow data.

```python
import os
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Ensure nids package is in path
root_dir = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(root_dir / 'src'))

from nids.data.loader import load_dataset_from_csv
from nids.data.audit import audit_dataframe
from nids.data.cleaning import clean_dataframe
from nids.data.label_mapping import map_labels, TARGET_CLASSES
```

### 1. Ingest Demonstration Dataset

```python
demo_path = root_dir / 'data' / 'demo' / 'cicids2017_demo.csv'
df = pd.read_csv(demo_path)
print(f"Loaded demo shape: {df.shape}")
df.head()
```

### 2. Dataset Audit: Missing, Infinite, and Duplicate Records

```python
audit_res = audit_dataframe(df)
print(f"Total Rows: {audit_res['total_rows']:,}")
print(f"Duplicates: {audit_res['duplicate_rows']:,}")
print(f"Missing Values: {audit_res['total_missing_values']:,}")
print(f"Infinite Values: {audit_res['total_infinite_values']:,}")
print("Label Distribution:", audit_res['label_distribution'])
```

### 3. Four-Class Target Mapping

```python
df_cleaned, _ = clean_dataframe(df, drop_duplicates=True)
df_mapped, map_rep = map_labels(df_cleaned)
print("Mapped Distribution:", map_rep['final_distribution'])

# Visualize 4-Class Distribution
plt.figure(figsize=(8, 4))
sns.countplot(data=df_mapped, x='Label', palette=['#10B981', '#EF4444', '#F59E0B', '#8B5CF6'])
plt.title('4-Class Network Traffic Distribution (CICIDS2017 Sample)', fontweight='bold')
plt.xlabel('Security Class')
plt.ylabel('Flow Count')
plt.tight_layout()
plt.show()
```

### 4. Correlation and Feature Distribution Analysis

```python
num_cols = df_mapped.select_dtypes(include=[np.number]).columns[:10]
plt.figure(figsize=(10, 8))
sns.heatmap(df_mapped[num_cols].corr(), annot=True, fmt='.2f', cmap='Blues')
plt.title('Correlation Heatmap (Top Flow Features)', fontweight='bold')
plt.tight_layout()
plt.show()
```
